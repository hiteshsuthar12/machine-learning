```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Customer data
# [Annual Income, Spending Score]
X = np.array([
    [20, 20],
    [22, 25],
    [25, 22],
    [24, 28],

    [70, 75],
    [72, 78],
    [75, 72],
    [78, 76],

    [40, 40],
    [42, 44],
    [45, 42],
    [43, 46]
])

print("Dataset:")
print(X)
```

```python
# Create K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

# Get centroids
centers = kmeans.cluster_centers_

print("Cluster Labels:")
print(labels)

print("\nCluster Centroids:")
print(centers)
```

```python
# Plot
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    s=100
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    color='red',
    marker='X',
    s=200,
    label='Centroids'
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")
plt.legend()
plt.grid()
plt.show()
```
